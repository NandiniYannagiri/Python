#### Regular expressions (Regex) are patterns used to search, match, extract, split, and replace text. In Python, regex is provided by the built-in re module.

### Why Regex

```python
# Without Regex
# Validating an E-mail
email = input()
if "@" in email and "." in email:
    print("Valid")
```

```python
# With Regex
# Validating an E-mail

import re
email = input()
if re.match(r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$', email):
    print("Valid")
else:
    print("Invalid")
```

### Regex Functions

```python
re.match()
re.findall()
re.search()
re.finditer()
re.sub()
re.split()
re.fullmatch()
```

```python
import re
text = "Python"
print(re.match("Py", text))
```

```python
print(re.match("thon", text))
```

```python
import re
text = "I love Python"
print(re.search("Python", text))
```

```python
print(re.search("love", text))
```

```python
import re
text = "cat bat mat rat"
print(re.findall("at", text))
```

```python
text = "abc123xyz456"
print(re.findall(r"\d+", text))
```

```python
import re
text = "abc123xyz456"
for i in re.finditer(r"\d+", text):
    print(i.group(), i.start(), i.end())
```

```python
import re
text = "I like cats"
print(re.sub("cats", "dogs", text))
```

```python
text = "abc123xyz456"
print(re.sub(r"\d", "*", text))
```

```python
import re
text = "apple,banana;orange grape"
print(re.split(r"[,; ]", text))
```

```python
import re
print(re.fullmatch(r"\d+", "12345"))
```

### Common Real world examples


```python
## phone number valifdation
import re
number = "9876543210"
print(bool(re.fullmatch(r"\d{10}", number)))
```

```python
# E-Mail validation
import re
email = input("Enter Email: ")
pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
if re.fullmatch(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")
```

```python
# Password validation
import re
password = input("Enter Password: ")
pattern = r'^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[@#$%^&+=!]).{8,}$'
if re.fullmatch(pattern, password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
## Square Pattern
n = 4
for i in range(n):
    for j in range(n):
        print("*",end =" ")
    print()
```

```python
## Hollow Square
n = 4
for i in range(n):
    for j in range(n):
        if i == 0 or i== n-1 or j == 0 or j == n-1:
            print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```

```python
## Number Patterns in Square Shape
## 1
n = 4
for i in range(1,n+1):
    for j in range(1,n+1):
        print(i,end = " ")
    print()
print("_________________________")
## 2
n = 4
for i in range(1,n+1):
    for j in range(1,n+1):
        print(j,end = " ")
    print()

print("_________________________")
## 3
n = 4
num = 1
for i in range(1,n+1):
    for j in range(1,n+1):
        print(num,end = "  ")
        num += 1
    print()

print("_________________________")
## 4
n = 4
even = 2
odd = 1
for i in range(n):
    if i%2 != 0:
        for j in range(n):
            print(even, end = "   ")
            even +=2
    else:
        for j in range(n):
            print(odd, end= "   ")
            odd+=2
    print()
        
```

```python
## Right Angle Trianlge Patttern
n = 4
for i in range(n):
    for j in range(i+1):
        print("*",end = " ")
    print()
```

```python
## Number patterns in Right angle triangle Pattern
## 1
n = 4
for i in range(1,n+1):
    for j in range(i):
        print(i,end = " ")
    print()
print("_________________________")
## 2
n = 4
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end = " ")
    print()

print("_________________________")

## 3
n = 4
num = 1
for i in range(1,n+1):
    for j in range(1,i+1):
        print(num,end = " ")
        num+=1
    print()

print("_________________________")

## 4
n = 4
even = 2
odd = 1
for i in range(1,n+1):
    for j in range(1,i+1):
        if i%2==0:
            print(even,end = " ")
            even+=2
        else:
            print(odd,end = " ")
            odd+=2
    print()
```

```python
## Invert Right angle triangle
## 1
n = 4
for i in range(n):
    for j in range(n-i):
        print("*",end=" ")
    print()

print("_________________________")
## 2
n = 4
for i in range(1,n+1):
    for j in range(n-i+1):
        print(i,end=" ")
    print()

print("_________________________")
## 3
n = 4
for i in range(1,n+1):
    for j in range(1,n-i+2):
        print(j,end=" ")
    print()
print("_________________________")
## 4
n = 4
even = 2
odd = 1
for i in range(1,n+1):
    for j in range(n-i+1):
        if i%2==0:
            print(even,end=" ")
            even+=2
        else:
            print(odd,end=" ")
            odd+=2
    print()
```

```python
## Pyramid Patterns
##1
n = 4
for i in range(n):
    ## spaces
    for j in range(n-i):
        print(" ",end = " ")
    ## stars
    for j in range(2*i+1):
        print("*",end = " ")
    print()
    
print("_________________________")
print()
## 2
n = 4
num = 1
for i in range(n):
    ## spaces
    for j in range(n-i):
        print(" ",end = " ")
    ## stars
    for j in range(2*i+1):
        print(num,end = " ")
    num+=1
    print()
    
print("_________________________")
print()

## 3
n = 4
num = 1
for i in range(n):
    ## spaces
    for j in range(n-i):
        print(" ",end = " ")
    ## stars
    for j in range(2*i+1):
        print(num,end = " ")
        num+=1
    num = 1
    print()

print("_________________________")
print()

## Invert Pyramid
## 1
n = 4
for i in range(n):
    ## spaces
    for j in range(i+1):
        print(" ",end = " ")
    ## stars
    for j in range(2*n-(2*i+1)):
        print("*",end = " ")
    print()

print("_________________________")
print()
## 2
n = 4
for i in range(n):
    ## spaces
    for j in range(i+1):
        print(" ",end = " ")
    ## stars
    for j in range(2*n-(2*i+1)):
        print(num,end = " ")
    num += 1
    print()
print("_________________________")
print()
## 3
n = 4
num = 1
for i in range(n):
    ## spaces
    for j in range(i+1):
        print(" ",end = " ")
    ## stars
    for j in range(2*n-(2*i+1)):
        print(num,end = " ")
        num += 1
    print()
```

```python
## Rhombus Pattern
n = 4
for i in range(n):
    ## spaces
    for j in range(n-i):
        print(" ",end = " ")
    ## stars
    for j in range(2*i+1):
        print("*",end = " ")
    print()
n = n-1
for i in range(n):
    ## spaces
    for j in range(i+2):
        print(" ",end = " ")
    ## stars
    for j in range(2*n-(2*i+1)):
        print("*",end = " ")
    print()
```
